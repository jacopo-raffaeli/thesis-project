```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
```

```python
year = "2022"
month = "08"
day = "01"

base_data_dir = Path("..") / "data" / "raw" / "FBTP"
fbtp_fut_path = base_data_dir / "futures" / year / month
fbtp_ctd_path = base_data_dir / "ctd" / year / month

date = f"{year}_{month}_{day}"
ctd_filename = f"ctd_lob_freq_1s_{date}.parquet"
fut_filename = f"futures_lob_freq_1s_{date}.parquet"

fut_file = fbtp_fut_path / fut_filename
ctd_file = fbtp_ctd_path / ctd_filename
```

```python
fut_df = pd.read_parquet(fut_file, engine="fastparquet")
ctd_df = pd.read_parquet(ctd_file, engine="fastparquet")
```

```python
fut_df.head()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(fut_df.index, fut_df["MidPrice"], label="FBTP Mid Price")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(fut_df.index, fut_df["MidPrice"], label="FBTP Mid Price", color="red")
plt.plot(ctd_df.index, ctd_df["MidPrice"], label="FBTP CTD Mid Price", color="blue")
```
